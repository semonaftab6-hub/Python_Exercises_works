```python
# wap to count a particular character in a given string
i=0
count=0
data="Python Programming"
while(i<len(data)):
    if data[i]=="P":
        count=count+1
    i=i+1
print(count)
```

```python
#wap to remove  all vowels of any string
new_data=""
data="Python Programming"
vowels='a,e,i,o,u,A,E,I,O,U'
for i in data:
    if i not in  vowels:
    
        new_data=new_data+i
print(new_data)
```

```python
#wap to remove @ and print only remaining string using for loop
data="invertis@123"
newdata=""
remove="@"
for i in data:
    if i!=remove:
        newdata=newdata+i
print(newdata)
        
        
        
```
