```python
#EXERCISE4
employees={"Rahul":{"salary":60000,"department":"IT"},"Amit":{"salary":40000,"department":"HR"}}
for values in employees:
    if employees[values]["salary"]>=50000:
        if employees[values]["department"]=="IT":
            print("IT EMPLOYEE")
        elif employees[values]["department"]=="HR":
            print("HR EMPLOYEE")
        else:
            print("Senior Employee")
    else:
        print("Junior Employee")
        
```

```python
#EXERCISE5
data={"a":10,"b":15,"c":25,"d":30,"e":40}
for values in data:
    if data[values]%2==0 and data[values]>20:
        print("large even ")
    elif data[values]%2==0 and data[values]<=20:
        print("even number")
    else:
        print("odd")
```

```python
#EXERCISE7
students={"Rahul":{"attendance":85,"marks":70},"Amit":{"attendance":60,"marks":80},"Priya":{"attendance":90,"marks":35},"Neha":{"attendance":78,"marks":55}}
for values in students:
    if students[values]["attendance"]>=75:
        if students[values]["marks"]>=40:
            print("Eligible")
        else:
            print("Failed In Exam")
    else:
        print("Not Eligible")
        
```

```python
#EXERCISE8
accounts={"Rahul":25000,"Amit":5000,"Priya":0,"Neha":-2000}
for values in accounts:
    if accounts[values]>0:
        if accounts[values]>=10000:
            print("Premium Account")
        else:
            print("Regular Account")
    elif accounts[values]==0:
        print("No Balance")
    else:
        print("Overdraft")
```

```python
#EXERCISE9
employees={"Semon":{"salary":60000,"experience":6},"Aish":{"salary":50000,"experience":4},"Mary":{"salary":200000,"experience":1}}
for values in employees:
    if employees[values]["salary"]>=50000:
        if employees[values]["experience"]>=5:
            print("Senior")
        else:
            print("Mid-Level")
    elif employees[values]["salary"]<50000:
        if employees[values]["experience"]>=3:
            print("Experienced Junior")
        else:
            print("Fresher/Junior")
```

```python
#EXERCISE10
students={"Rahul":{"marks":85,"attendance":90},"Amit":{"marks":35,"attendance":80},"Priya":{"marks":65,"attendance":70},"Neha":{"marks":45,"attendance":85},"Rohit":{"marks":90,"attendance":60}}
for values in students:
    if students[values]["attendance"]<75:
        print(values ,"Not Eligible")
    elif students[values]["attendance"]>=75:
        if students[values]["marks"]>=80:
            print(values,"Distinction")
        elif students[values]["marks"]>=60:
            print(values,"First Division")
        elif students[values]["marks"]>=40:
            print(values,"Pass")
        else:
            print(values,"Fail")
        
```
