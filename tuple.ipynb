```python
#Tuple : paranthesis are commonly used to create tuple
data=(10,20,30,40,50)
print(type(data))
```

```python
#but paranthesis are actually not mandatory
#the comma , is what makes a tuple , not the paranthesis
data2=10,20,30,40,50
print(type(data2))
```

```python
data3=(10)
print(type(data3))
# Output is <class 'int'>
#why?
# Because(10) is not an tuple , it is simply a integer enclosed in paranthesis
#python treats it as :
# data3=10
```

```python
#how to create single element tuple ?
#you must use a comma
data4=(10,)
print(type(data4))
#remember this rule:
#code            type
# data=(10)      int
# data=(10,)     tuple
#  data=10,       tuple
# data=(10,20)     tuple
```

```python
#tuple indexing
#tuple are ordered and immutable
data=(10,20,30,40,50)
print(data[2]) # indexing is used to access individul elements from a tuple  
print(data[-1]) #negative indexing starts from -1 from right side
```

```python
#tuple allow duplicate elements
data=(10,20,30,40,40,50,60)
print(data)
```

```python
#tuple is immutable
#a tuple is immutable that means its elements cant be updated,changed,or deleted after creation
data=(10,20,30,40,50)
data[2]=80
print(data)
```
