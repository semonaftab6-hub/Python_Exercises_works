```python
#add()
data={10,23,43,22,65,78}
data.add(10)
print(data)
```

```python
#update()
data={10,23,43,22,65,78}
data.update([10,20,30])
print(data)
```

```python
#remove()
data={10,23,43,22,65,78}
data.remove(65)
print(data)
data.remove(55)  
print(data)
```

```python
#discard()
data={10,23,43,22,65,78}
data.discard(55)
print(data)
```

```python
# difference between remove and discard is that :
# remove will give an key error if the element given in PARENTHESIS to be removed is not present in the set
# discard will not result an error ,,it will return the set unchanged
```

```python
#pop()
data={10,23,43,22,65,78}
data.pop() 
print(data)
```
