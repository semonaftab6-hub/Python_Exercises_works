```python
#REAL TIME PROJECT1:STUDENT MANAGEMENT SYSTEM
class Students:
    def __init__(self,name,marks):
        self.name=name
        self.marks=marks

    def showdetails(self):
        total=sum(self.marks)
        average=total/len(self.marks)
        print("Name:",self.name)
        print("Marks:",self.marks)
        print("Total Marks",total)
        print("Average:",average)
        if average>=90:
            grade="A"
        elif average>=80:
            grade="B"
                
        elif average>=60:
            grade="C"
               
        else:
            grade="F"
        print("overall grade",grade)
        
        
object1=Students("semon",[45,67,89])
object1.showdetails()
print("="*20)
object1=Students("aish",[87,90,82])
object1.showdetails()
```

```python
#REALTIME PROJECT3>>EMPLOYEE PAYROLL SYSTEM
class EmployeePayrollSystem:
    def __init__(self,employee_ID,name,basic_salary):
        self.empid=employee_ID
        self.name=name
        self.basic_salary=basic_salary
    def calculate_salary(self):
        print("employee_id:",self.empid)
        print("employee_name:",self.name)
        print("employee_basic_salary:",self.basic_salary)
        hra=self.basic_salary*20/100
        da=self.basic_salary*10/100
        pf=self.basic_salary*12/100
        net_salary=self.basic_salary+hra+da-pf
        print("netsalary:",net_salary)
e1=EmployeePayrollSystem(121,"semon",45000)
e1.calculate_salary()
```
