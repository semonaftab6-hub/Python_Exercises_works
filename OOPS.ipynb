```python
#REAL TIME PROJECT1:STUDENT MANAGEMENT SYSTEM
class Students:
    def __init__(self,name,marks):
        self.name=name
        self.marks=marks

    def showdetails(self):
        total=sum(self.marks)
        average=total/len(self.marks)
        print("Name:",self.name)
        print("Marks:",self.marks)
        print("Total Marks",total)
        print("Average:",average)
        if average>=90:
            grade="A"
        elif average>=80:
            grade="B"
                
        elif average>=60:
            grade="C"
               
        else:
            grade="F"
        print("overall grade",grade)
        
        
object1=Students("semon",[45,67,89])
object1.showdetails()
print("="*20)
object1=Students("aish",[87,90,82])
object1.showdetails()
```

```python
#REALTIME PROJECT3>>EMPLOYEE PAYROLL SYSTEM
class EmployeePayrollSystem:
    def __init__(self,employee_ID,name,basic_salary):
        self.empid=employee_ID
        self.name=name
        self.basic_salary=basic_salary
    def calculate_salary(self):
        print("employee_id:",self.empid)
        print("employee_name:",self.name)
        print("employee_basic_salary:",self.basic_salary)
        hra=self.basic_salary*20/100
        da=self.basic_salary*10/100
        pf=self.basic_salary*12/100
        net_salary=self.basic_salary+hra+da-pf
        print("netsalary:",net_salary)
e1=EmployeePayrollSystem(121,"semon",45000)
e1.calculate_salary()
```

```python
#single inheritance
class BankAccount(): #class 1 -> parent class
    def __init__(self,AccountHolderName,balance):
        self.AccountHolderName=AccountHolderName
        self.balance=balance
        
    def showBalance(self):
        print("Account Holder Name:",self.AccountHolderName)
        print("Balance:",self.balance)
        
class SavingAccount(BankAccount): #class 2->child class 
    def addInterestRate(self,rate):
        interest=self.balance*rate/100
        self.balance+=interest
        print("interest added",interest)
s1=SavingAccount("Akhilesh",20000) 
s1.showBalance()
print("="*25)
s1.addInterestRate(8.0)
s1.showBalance()
```

```python
#Multilevel Inheritance
class Employee:
    def __init__(self,Name,Salary):
        self.name=Name
        self.salary=Salary
    def showEmpDetails(self):
        print("Name :",self.name)
        print("Salary:",self.salary)
        
class HR(Employee):
    def calculatePF(self,pf):
        finalSalary=self.salary-pf
        print("Final Salary After PF:",finalSalary)
        
class Accounts(Employee):
    def calculatePF(self,pf):
        finalSalary=self.salary-pf
        print("Final Salary After PF:",finalSalary)
        
h1=HR("Amit",20000)
h1.calculatePF(1200)
a1=Accounts("Pawan",10000)
a1.calculatePF(1000)
a1.showEmpDetails() #using object of child class to inherit showEmpDetails function of the parent class
h1.showEmpDetails()#using object of child class to inherit showEmpDetails function of the parent class
```

```python
#constructor Overriding
class Employee():
    def __init__(self,Name):
        self.Name=Name
        print("Employee Kaa Constructor Call Ho gya")
class Developer(Employee):
    def __init__(self,Name,language):
        super().__init__(Name)#Jab child class (Developer) me bhi __init__ likh diya jaye, toh python confuse ho jata hai ki kiska constructor chale. Isliye hum super().__init__(Name) likhte hain taaki parent class ka data bhi initialize ho jaye aur error na aaye.
        self.language=language
        print("Developer Ka constructor call ho gya")
    def showDetails(self):
        print("Employee Name:",self.Name)
        print("Language:",self.language)
d1=Developer("Amit","Python")
d1.showDetails()
#issubclass()
#isinstance()
print(isinstance(d1,Developer))  #isinstance(obj, class): Ye check karta hai ki kya koi object kisi specific class ka bana hua hai ya nahi. 
print(issubclass(Developer,Employee))#issubclass(child, parent): Ye check karta hai ki kya ek class, dusri class ki child class hai ya nahi.
print(issubclass(Employee,Developer))
```

```python
#getter and setter methods
class Employee():
    def __init__(self,name,profile,salary):
        self.__name=name
        self.__profile=profile
        self.__salary=salary
    def setName(self,name):
        self.__name=name
    def setProfile(self,profile):
        self.__profile=profile
    def setSalary(self,salary):
        self.__salary=salary
    def getName(self):
        return self.__name
    def getProfile(self):
        return self.__profile
    def getSalary(self):
        return self.__salary
    
    
e1=Employee("semon","HR",34000)
e1.setName("semon aftab")
print(e1.getName())
e1.setProfile("IT")
print(e1.getProfile())
e1.setSalary(50000)
print(e1.getSalary())
```

```python
from abc import ABC,abstractmethod
class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

class Dog(Animal):
    def sound(self):#method overriding
        print("DOG BARKS")

class Cat(Animal):
    def sound(self):
        print("cat meows")#method overriding

d1=Dog()
d1.sound()

c1=Cat()
c1.sound()
```

```python
from abc import ABC,abstractmethod
class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass
class Dog(Animal):
    pass
class Cat(Animal):
    pass
d1=Dog()
d1.sound()
c1=Cat()
c1.sound()
    
```
