```python
#ex6
def calculate_avg(*marks):
    if len(marks)<0:
        return "no marks"
    for ex in marks:
        if not isinstance(ex,(int,float)) or isinstance(ex,bool):#agar esa na ho
            return "error marks not in int or float"
        if ex<0 or ex>100:#(1-100)
            return "error:marks must be between 0 and 100"
    return sum(marks)/len(marks)
        
print(calculate_avg(55,66,78,90))#avg of these data using loop
print(calculate_avg(55,66,78,"hello",90,89.0,True,False))#marks should be strictly int,float for that we will give a filter
print(calculate_avg(55,66,78,90,89.0,True,False))#it will give no error because True=1,False=0,but 
print(calculate_avg(55,66,78,90,89.0,600))
print(calculate_avg(55,66,78,90,89.0))
```

```python
students=[
    {'name': 'Aman', 'marks':[80,75,90] },
    {'name': 'Riya', 'marks':[35,40,30] },
    {'name': 'Rahul', 'marks':[90,85,95] },
    {'name': 'Neha', 'marks':[60,70,65] }
]
def student_result(students):
    for student in students:
        name=student["name"]
        marks=student["marks"]
        avg=sum(marks)/len(marks)
        if avg>=40:
            status="Pass"
        else:
            status="Fail"
        yield{
           "name":name,
           "average":avg,
           "status":status
        }
ex=student_result(students)
print(next(ex))
print(next(ex))
print(next(ex))
print(next(ex))
```

```python
#REAL TIME PROJECT1:STUDENT MANAGEMENT SYSTEM
class Students:
    def __init__(self,name,marks):
        self.name=name
        self.marks=marks

    def showdetails(self):
        total=sum(self.marks)
        average=total/len(self.marks)
        print("Name:",self.name)
        print("Marks:",self.marks)
        print("Total Marks",total)
        print("Average:",average)
        if average>=90:
            grade="A"
        elif average>=80:
            grade="B"
                
        elif average>=60:
            grade="C"
               
        else:
            grade="F"
        print("overall grade",grade)
        
        
object1=Students("semon",[45,67,89])
object1.showdetails()
print("="*20)
object1=Students("aish",[87,90,82])
object1.showdetails()
```

```python
#REALTIME PROJECT3>>EMPLOYEE PAYROLL SYSTEM
class EmployeePayrollSystem:
    def __init__(self,employee_ID,name,basic_salary):
        self.empid=employee_ID
        self.name=name
        self.basic_salary=basic_salary
    def calculate_salary(self):
        print("employee_id:",self.empid)
        print("employee_name:",self.name)
        print("employee_basic_salary:",self.basic_salary)
        hra=self.basic_salary*20/100
        da=self.basic_salary*10/100
        pf=self.basic_salary*12/100
        net_salary=self.basic_salary+hra+da-pf
        print("netsalary:",net_salary)
e1=EmployeePayrollSystem(121,"semon",45000)
e1.calculate_salary()
```
