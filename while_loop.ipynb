```python
# printing numbers from 1 to 10 using while loop
i=1 #initialisation 
while(i<=10): 
    print(i)
    i=i+1
    

```

```python
# printing even numbers from 1 to 20
i=1
while(i<=20):
    if i%2==0:
        print(i)
    i=i+1
```

```python
#printing odd numbers from 1 to 20
i=1
while(i<=20):
    if(i%2!=0):
        print(i)
    i=i+1
```

```python
#find the sum of numbers from 1 to 10
i=1
sum=0
while(i<=10):
    sum=sum+i
    i=i+1
print(sum)
    
```

```python
#finding sum of  even numbers from 1 to 100
i=1
sum=0
while(i<=100):
    if(i%2==0):
        sum=sum+i
    i=i+1
print(sum)
```

```python
#finding sum of  odd numbers from 1 to 100
i=1
sum=0
while(i<=100):
    if(i%2!=0):
        sum=sum+i
    i=i+1
print(sum)
```

```python
##find the multiplication table of 7
i=1
while (i<=10):
    print("7 x",i,"=",7*i)
    i=i+1
```

```python
#print the numbers from 1 to 100 that are divisible by both 3 and 5
i=1
while(i<=100):
   if i%3==0 and i%5==0:
       print(i)
   i=i+1
```

```python
# count even and odd numbers between 1 and 50
e_count=0
o_count=0
i=1
while(i<=50):
    if (i%2==0):
        e_count=e_count+1
    else:
        o_count=o_count+1
    i=i+1
print(e_count)
print(o_count)
```

```python
#printing largest no from a list
list_1=[10,40,14,75,43,90]
largest=list_1[0]
i=0
while(i<=5):
    if list_1[i]>largest:
        largest=list_1[i]
    i=i+1
print(largest)
```

```python
#printing smallest no. from alist
list_2=[10,40,14,75,43,90]
smallest=list_2[0]
i=0
while(i<=5):
    if list_2[i]<smallest:
        smallest=list_2[i]
    i=i+1
print(smallest)
    
```

```python
#reversing a list 
list_2=[10,40,14,75,43,90]
i=0
while(i<=5):
    print(list_2[::-1])
i=i+1
```
