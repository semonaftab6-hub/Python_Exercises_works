```python
#ex2
sub1=int(input("enter marks1:"))
sub2=int(input("enter marks2:"))
sub3=int(input("enter marks3:"))
if sub1>=40 and sub2>=40 and sub3 >=40:
    percentage=((sub1+sub2+sub3)/300*100)
    if percentage>=75:
        print("Distinction",percentage,"%")
    elif percentage>=60:
        print("First Division",percentage,"%")
    elif percentage>=50:
        print("Second Division",percentage,"%")
    else:
        print("pass",percentage,"%")
else:
    print("Fail",percentage,"%")
    
```
